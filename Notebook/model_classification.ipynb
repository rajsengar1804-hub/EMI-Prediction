```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)
```

```python
numerical_with_nan = [
    feature
    for feature in X_train.columns
    if X_train[feature].isnull().sum() > 0
    and X_train[feature].dtype != "object"
]

print(numerical_with_nan)
```

```python
for feature in numerical_with_nan:
    print(X_train[feature].dtypes)
```

```python
from sklearn.impute import SimpleImputer

num_imputer= SimpleImputer(strategy="median")

X_train[numerical_with_nan] = num_imputer.fit_transform(
    X_train[numerical_with_nan]
)

X_test[numerical_with_nan] = num_imputer.transform(
    X_test[numerical_with_nan]
)
```

```python
X_train[numerical_with_nan].isnull().sum()
```

```python
X_test[numerical_with_nan].isnull().sum()
```

```python
cat_imputer = SimpleImputer(strategy="most_frequent")

X_train[["education"]] = cat_imputer.fit_transform(
    X_train[["education"]]
)

X_test[["education"]] = cat_imputer.transform(
    X_test[["education"]]
)
```

```python
X_train.isnull().sum()
```

```python
X_test.isnull().sum()
```

```python
yeo_johnson_features = [
    'age',
    'monthly_salary',
    'years_of_employment',
    'monthly_rent',
    'college_fees',
    'travel_expenses',
    'groceries_utilities',
    'other_monthly_expenses',
    'current_emi_amount',
    'credit_score',
    'bank_balance',
    'emergency_fund',
    'requested_amount',
    'requested_tenure'
]
```

```python
import numpy as np
```

```python
X_train["school_fees"] = np.sqrt(
    X_train["school_fees"]
)

X_test["school_fees"] = np.sqrt(
    X_test["school_fees"]
)
```

```python
from sklearn.preprocessing import PowerTransformer

pt_cls = PowerTransformer(
    method="yeo-johnson",
    standardize=False
)

X_train[yeo_johnson_features] = pt_cls.fit_transform(
    X_train[yeo_johnson_features]
)

X_test[yeo_johnson_features] = pt_cls.transform(
    X_test[yeo_johnson_features]
)
print("done")
```

```python
categorical_features = [
    feature
    for feature in X_train.columns
    if X_train[feature].dtype == "object"
]

print(categorical_features)
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder_cls = OneHotEncoder(
    drop="first",
    handle_unknown="ignore",
    sparse_output=False
)
print("done")
```

```python
X_train_encoded = encoder_cls.fit_transform(
    X_train[categorical_features]
)

X_test_encoded = encoder_cls.transform(
    X_test[categorical_features]
)
print("done")
```

```python
encoded_columns = encoder_cls.get_feature_names_out(
    categorical_features
)

X_train_encoded = pd.DataFrame(
    X_train_encoded,
    columns=encoded_columns,
    index=X_train.index
)

X_test_encoded = pd.DataFrame(
    X_test_encoded,
    columns=encoded_columns,
    index=X_test.index
)
print("done")
```

```python
X_train = X_train.drop(
    columns=categorical_features
)

X_test = X_test.drop(
    columns=categorical_features
)
```

```python
X_train.shape
```

```python
X_train_final = pd.concat(
    [X_train, X_train_encoded],
    axis=1
)

X_test_final = pd.concat(
    [X_test, X_test_encoded],
    axis=1
)
```

```python
print(X_train_final.shape)
print(X_test_final.shape)
```

```python
import pandas as pd
import numpy as np

from sklearn.feature_selection import VarianceThreshold
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.pipeline import Pipeline
```

```python
X_train_final.info()
```

```python
variance_selector = VarianceThreshold(threshold=0.01)

X_train_var = variance_selector.fit_transform(X_train_final)
X_test_var = variance_selector.transform(X_test_final)

variance_features = X_train_final.columns[
    variance_selector.get_support()
]

X_train_var = pd.DataFrame(
    X_train_var,
    columns=variance_features,
    index=X_train_final.index
)

X_test_var = pd.DataFrame(
    X_test_var,
    columns=variance_features,
    index=X_test_final.index
)

print("Features before variance:", X_train_final.shape[1])
print("Features after variance:", X_train_var.shape[1])
```

```python
X_train_final.columns
```

```python
corr_matrix = X_train_var.corr().abs()

upper_triangle = corr_matrix.where(
    np.triu(
        np.ones(corr_matrix.shape),
        k=1
    ).astype(bool)
)

correlated_features = [
    column
    for column in upper_triangle.columns
    if any(upper_triangle[column] > 0.90)
]

print("Highly correlated features:", len(correlated_features))
print(correlated_features)
```

```python
X_train_corr = X_train_var.drop(
    columns=correlated_features
)

X_test_corr = X_test_var.drop(
    columns=correlated_features
)

print("Features after correlation:", X_train_corr.shape[1])
```

```python
rf = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1,
    class_weight="balanced"
)

rf.fit(X_train_corr, y_train)
```

```python
feature_importance = pd.DataFrame({
    "feature": X_train_corr.columns,
    "importance": rf.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="importance",
    ascending=False
)

print(feature_importance)
```

```python
ranked_features = feature_importance["feature"].tolist()

print(ranked_features)
```

```python
k_values = [5, 10, 15, 20, 25, 30]

k_values = [
    k for k in k_values
    if k <= len(ranked_features)
]

print("K values:", k_values)
```

```python
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
results = []

for k in k_values:

    selected_features = ranked_features[:k]

    X_k = X_train_corr[selected_features]

    model = RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        n_jobs=-1
    )

    scores = cross_val_score(
        model,
        X_k,
        y_train,
        cv=cv,
        scoring="f1_macro",
        n_jobs=-1
    )

    results.append({
        "K": k,
        "Mean_Macro_F1": scores.mean(),
        "Std_Macro_F1": scores.std()
    })

results_df = pd.DataFrame(results)

print(results_df)
```

```python
best_k = int(
    results_df.loc[
        results_df["Mean_Macro_F1"].idxmax(),
        "K"
    ]
)

print("Best K:", best_k)
```

```python
best_features = ranked_features[:best_k]

print("Selected features:")

for i, feature in enumerate(best_features, 1):
    print(i, feature)
```

```python
best_k = 15

best_features = ranked_features[:best_k]

X_train_final = X_train_corr[best_features]
X_test_final = X_test_corr[best_features]

print("Best K:", best_k)
print("X_train_final:", X_train_final.shape)
print("X_test_final :", X_test_final.shape)

print("\nSelected Features:")
print(best_features)
```

```python
import pandas as pd

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier
from xgboost import XGBClassifier
```

```python
from sklearn.model_selection import train_test_split

X_train_base, X_val, y_train_base, y_val = train_test_split(
    X_train_final,
    y_train,
    test_size=0.20,
    stratify=y_train,
    random_state=42
)

print("X_train_base:", X_train_base.shape)
print("X_val:", X_val.shape)
print("X_test_final:", X_test_final.shape)

print("y_train_base:", y_train_base.shape)
print("y_val:", y_val.shape)
print("y_test:", y_test.shape)
```

```python
print("Training distribution:")
print(y_train_base.value_counts(normalize=True))

print("\nValidation distribution:")
print(y_val.value_counts(normalize=True))

print("\nTest distribution:")
print(y_test.value_counts(normalize=True))
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

y_train_base_encoded = label_encoder.fit_transform(y_train_base)
y_val_encoded = label_encoder.transform(y_val)

print("Classes:", label_encoder.classes_)
```

```python
models = {

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        n_jobs=-1
    ),

    "LightGBM": LGBMClassifier(
        random_state=42,
        n_jobs=-1,
        verbosity=-1
    ),

    "CatBoost": CatBoostClassifier(
        iterations=200,
        random_state=42,
        verbose=False
    ),

    "XGBoost": XGBClassifier(
        n_estimators=200,
        random_state=42,
        n_jobs=-1,
        eval_metric="mlogloss"
    )
}
```

```python
baseline_results = []

for name, model in models.items():

    print(f"Training {name}...")

    model.fit(
        X_train_base,
        y_train_base_encoded
    )

    y_val_pred = model.predict(X_val)

    baseline_results.append({
        "Model": name,
        "Accuracy": accuracy_score(
            y_val_encoded,
            y_val_pred
        ),
        "Balanced Accuracy": balanced_accuracy_score(
            y_val_encoded,
            y_val_pred
        ),
        "Macro Precision": precision_score(
            y_val_encoded,
            y_val_pred,
            average="macro",
            zero_division=0
        ),
        "Macro Recall": recall_score(
            y_val_encoded,
            y_val_pred,
            average="macro",
            zero_division=0
        ),
        "Macro F1": f1_score(
            y_val_encoded,
            y_val_pred,
            average="macro",
            zero_division=0
        )
    })

baseline_results_df = pd.DataFrame(baseline_results)

baseline_results_df.sort_values(
    by="Macro F1",
    ascending=False
)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

xgb_model = models["XGBoost"]

y_val_pred_xgb = xgb_model.predict(X_val)

print("Classification Report:")
print(
    classification_report(
        y_val_encoded,
        y_val_pred_xgb,
        target_names=label_encoder.classes_,
        zero_division=0
    )
)

print("\nConfusion Matrix:")
print(
    confusion_matrix(
        y_val_encoded,
        y_val_pred_xgb
    )
)
```

## Because the primary objective of the EMI eligibility system is to identify high-risk applicants and minimize the number of high-risk applicants incorrectly classified as lower-risk, High_Risk Recall is treated as the primary evaluation metric. Balanced XGBoost was therefore preferred over SMOTE + XGBoost because it achieved 78% High_Risk Recall compared with 41%

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

y_train_encoded = label_encoder.fit_transform(y_train)
y_test_encoded = label_encoder.transform(y_test)

print("Classes:")
print(label_encoder.classes_)
```

```python
from sklearn.utils.class_weight import compute_sample_weight

final_sample_weights = compute_sample_weight(
    class_weight="balanced",
    y=y_train_encoded
)

print("Sample weights shape:", final_sample_weights.shape)
```

```python
from xgboost import XGBClassifier

final_xgb = XGBClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1,
    eval_metric="mlogloss"
)
```

```python
final_xgb.fit(
    X_train_final,
    y_train_encoded,
    sample_weight=final_sample_weights
)

print("Final model trained successfully.")
```

```python
y_test_pred = final_xgb.predict(
    X_test_final
)
```

```python
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

print("FINAL TEST RESULTS")
print("===================")

print("Accuracy:",
      accuracy_score(
          y_test_encoded,
          y_test_pred
      ))

print("Balanced Accuracy:",
      balanced_accuracy_score(
          y_test_encoded,
          y_test_pred
      ))

print("Macro Precision:",
      precision_score(
          y_test_encoded,
          y_test_pred,
          average="macro",
          zero_division=0
      ))

print("Macro Recall:",
      recall_score(
          y_test_encoded,
          y_test_pred,
          average="macro",
          zero_division=0
      ))

print("Macro F1:",
      f1_score(
          y_test_encoded,
          y_test_pred,
          average="macro",
          zero_division=0
      ))
```
